# enumerate()
Utile quando in un ciclo iterativo vuoi sempre sapere l'indice in cui ti trovi, in quanto restituisce sia l'indice corrente, che aumenta ad ogni elemento, sia l'elemento della lista, tuple, set, dizionario, range, stringa ecc che stai iterando, per non dovre usare .index() che è spesso scomodo quando più elementi si ripetono

```python
frutta = ["mele", "pere", "banane", "pere"]
for indice, elemento in enumerate(frutta):
    print(f" {indice} -> {elemento}")
```

# Counter()
Proveniente dalla libreria collienctions, permette di prendere un iterabile e contare quante volte di ripete ogni singolo elemento, memorizzando in una variabile un dizionario con l'elemento come chiave e il numero di volte che si ripete come valore. Utile anche ad estrarre il numero di elementi diversi, con len(Counter(iterable))

```python
from collections import Counter

frutta = ("mela", "mela", "pera", "pera", "banana", "pesca")
num_frutta = Counter(frutta)
for key in num_frutta.keys():
    print(f"{key} -> {num_frutta[key]}")
```
