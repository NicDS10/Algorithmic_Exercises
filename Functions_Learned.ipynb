# enumerate()
Utile quando in un ciclo iterativo vuoi sempre sapere l'indice in cui ti trovi, in quanto restituisce sia l'indice corrente, che aumenta ad ogni elemento, sia l'elemento della lista, tuple, set, dizionario, range, stringa ecc che stai iterando, per non dovre usare .index() che è spesso scomodo quando più elementi si ripetono

```python
frutta = ["mele", "pere", "banane", "pere"]
for indice, elemento in enumerate(frutta):
    print(f" {indice} -> {elemento}")
```

# collections library
Il più importante elemento, Counter(), permette di prendere un iterabile e contare quante volte di ripete ogni singolo elemento, memorizzando in una variabile un dizionario con l'elemento come chiave e il numero di volte che si ripete come valore. Utile anche ad estrarre il numero di elementi diversi, con len(Counter(iterable)). Altri metodi, come deque, permettono di aggiungere elementi all'inizio o alla fine di un iterabile senza un for e quindi n². In generale, utile per manipolare gli iterabili senza andare in TLE.

```python
from collections import Counter

frutta = ("mela", "mela", "pera", "pera", "banana", "pesca")
num_frutta = Counter(frutta)
for key in num_frutta.keys():
    print(f"{key} -> {num_frutta[key]}")
```

# sys library
Dalla libreria sys, che interagisce direttamente con l'interprete di Python, e permette di manipolare i tre canali di input, con stdin, di output, con stdout e di errore, con stderr. Alcuni metodi importanti sono sys.stdin.read(), che unito a .split() prende l'input tutto insieme in modo molto veloce e lo restituisce spezzato nei vari elementi; simile è sys.stdin.readline(). Permette inoltre di modificare lo standard input e oitup con stdin e stdout appunto, associandoli a file da cui prendere o salvare valori con open(). Con sys.setrecursionlimit() aumentiamo il limite a cui si blocca l'algoritmo con grandi input.

```python
import sys

sys.stdin = open('input.txt', 'r')
sys.stdout = open('output.txt', 'w')

input = list(map(int, sys.stdin.read().split()))
```

# map()
Prende tutti i valori di un iterabile o in generale un insieme di elementi e applica una funzione a ciascuno.

```python
import sys

input = list(map(int, sys.stdin.read().split()))
```

# .sort()
Permette di ordinare un iterabile secondo un certo ordine, per esempio dei numeri in ordine crescente

```python
list = [5, 7, 8, 4, 6, 8, 0]
list.sort()
for num in list:
    print(num, end=" ")
```

# @cache
Un trucchetto per la programmazione dinamica, in particolare la memoization. Permette di salvare i risultati delle funzioni in automatico, riutilizzandole quando necessario.

```python
from functools import cache

@cache      # Python crea automaticamente un dizionario nascosto per salvare i risultati
def fib_super_memo(n):
    if n == 1 or n == 2:
        return 1
    return fib_super_memo(n - 1) + fib_super_memo(n - 2)

fib_super_memo(100)
```
